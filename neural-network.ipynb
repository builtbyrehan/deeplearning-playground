```python
def relu(z):
    if z > 0:
        return z
    else: 
        return 0

x1 = 2
x2 = 3
z1 = (x1 * 0.5) + (x2 * 0.8) + 1 # neuron 1
z2 = (x1 * 0.2) + (x2 * 0.4) + 0.5 # neuron 2
z3 = (x1 * 0.9) + (x2 * 0.1) + 0.2

print("Neuron 1: ",z1)
print("Neuron 2: ",z2)
print("Neuron 2: ",z3)

output_1 = relu(z1)
output_2 = relu(z2)
output_3 = relu(z3)

print(output_1," -> ",output_2," -> ", output_3)

```

```python
# now these 3 neurons will become the input of the next neuron.

# weights of the next neuron

w1,w2,w3 = 0.3, 0.6, 0.2
bias = 0.5
z_output = (output_1 * w1) + (output_2 * w2) + (output_3 * w3) + bias
print(z_output)
```

```python
# now we will implement the sigmoid function and will learn that how we may squeeze a bigger value into small values anywhere between 0 and 1

import math

def sigmoid(z):
    return (1/1 + math.exp(-z))

prediction = sigmoid(z_output)
print(prediction)
```

### Loss Function

#### Binary-Cross-Entropy

loss = -[y log(p) + (1 - y) log(1 - p)]

```python
import math

def binary_cross_entropy(target, prediction):
    loss = -(target * math.log(prediction) +
             (1 - target) * math.log(1 - prediction))
    
    return loss

target = 1
prediction = 0.87

loss = binary_cross_entropy(target, prediction)

print(loss)
```

### **Gradient Descent**

##### Gradient Descent is used to adjust the model’s weights so the loss becomes smaller and predictions become more accurate.

```python
# Input and correct answer
x = 2
target = 10

# Starting weight
weight = 1.0

# How fast we update the weight
learning_rate = 0.01

for epoch in range(10):

    # 1. Make prediction
    prediction = x * weight

    # 2. Calculate loss
    loss = (target - prediction) ** 2

    # 3. Calculate gradient
    gradient = 2 * (prediction - target) * x

    # 4. Update weight
    weight = weight - learning_rate * gradient

    print(
        "Prediction:", prediction,
        "Loss:", loss,
        "Weight:", weight
    )
```
